```python
import csv
from pathlib import Path
from collections import Counter

PROJECT_ROOT = Path.cwd().parent
CATALOG_PATH = PROJECT_ROOT / "data" / "processed" / "catalog.csv"


def main():
    if not CATALOG_PATH.exists():
        raise FileNotFoundError(
            "catalog.csv not found. Run prepare_dataset.py first."
        )

    with CATALOG_PATH.open(
        "r", encoding="utf-8", newline=""
    ) as file:
        records = list(csv.DictReader(file))

    print(f"Total records: {len(records)}")

    counts = Counter(row["category"] for row in records)

    print("\nImages per category:")
    for category, count in sorted(counts.items()):
        print(f"  {category}: {count}")

    print("\nSample records:")
    for record in records[:5]:
        record["image_id"] = int(record["image_id"])
        record["width"] = int(record["width"])
        record["height"] = int(record["height"])
        print(record)


if __name__ == "__main__":
    main()
```
